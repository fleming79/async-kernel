# Client (Notebook)

A client communicates with a kernel either locally in the same process, or remotely using sockets. The interface provides connections to the kernel, to which a client can connect.

## LocalClient

A local client can be used in the same process to communicate with the kernel. A local client is used in JupyterLite to pass comms to the Jupyterfrontend. It may also be convenient to use a LocalClient for running tests.

```python
# task
# We run this code block as a task to avoid deadlock as the client execute would be queued behind this request.
from async_kernel.messaging import LocalClient

async with LocalClient().start() as client:
    reply = await client.execute("a=1+1", user_expressions={"a": "a"})
reply
```

## ZMQClient

A `ZMQClient` communicates using zmq sockets meaning it can communicate with a kernel on the network.

Let's start a new kernel and get the kernel info.

```python
from async_kernel.messaging.zmq import ZMQClient

client = ZMQClient()
async with client.subprocess_kernel(name="my_kernel"):
    reply = await client.kernel_info()
reply
```
